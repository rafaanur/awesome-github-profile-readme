```python
# Kasus: Ibu Guru punya 25 buah permen, ingin dibagikan rata ke 4 siswa.
jumlah_permen = 25
jumlah_siswa = 4

permen_per_siswa = jumlah_permen // jumlah_siswa
sisa_permen = jumlah_permen % jumlah_siswa

print(f"Setiap siswa mendapat : {permen_per_siswa} permen")
print(f"Sisa permen di toples : {sisa_permen} permen")
```
