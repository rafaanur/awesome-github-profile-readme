```python
%%writefile luas_lingkaran.cpp
#include <iostream>
using namespace std;

int main()
{
    float jari_jari, luas;
    float phi = 3.14;

    cout << "Program Aplikasi Menghitung Luas Lingkaran" << endl;
    cout << "Masukkan jari-jari: ";
    cin >> jari_jari;

    luas = phi * jari_jari * jari_jari;

    cout << "Luas lingkaran"<< luas << endl;

    return 0;
}
```
