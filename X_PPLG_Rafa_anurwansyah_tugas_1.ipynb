Tugas 1 praktik membuat tipe data dan variabel

nama kelompok:iq+ketenangan
nama anggota:
1.rafa anurwansyah
2.syafik hakim
3.habibi hamka
4.fiqih ramadhan
5.satria apriliansyah

```python
%%writefile praktik_variabel.cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    // Deklarasi Variabel dan Tipe Data
    string namaBarang = "iPhone Kupih";
    int jumlah = 2;
    float harga = 7500000.50;
    char kategori = 'A';
    bool tersedia = true;

    // Menampilkan Output ke Layar
    cout << "=== DATA INVENTARIS ===" << endl;
    cout << "iPhone Kupih: " << namaBarang << endl;
    cout << "Kategori    : " << kategori << endl;
    cout << "Jumlah      : " << jumlah << endl;
    cout << "75000000.50 : Rp " << harga << endl;
    cout << "Tersedia    : " << tersedia << " (1 = Ya, 0 = Tidak)" << endl;
    cout << "=======================" << endl;

    return 0;
}
```

```python
!g++ praktik_variabel.cpp -o praktik_variabel

!./praktik_variabel
```
